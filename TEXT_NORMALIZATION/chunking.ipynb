## Fixed Size Chunking

```python
with open("Data.txt","r") as file:
    data = file.read()
    chunks = []
    fixed_size = 20
    for index in range(0, len(data), fixed_size): # 0 20
        chunk = data[index:index + fixed_size] #data[0:0+20] data[0:20]  #data[20:20+20]
        # print(chunk,len(chunk))
        chunks.append(chunk)
print(chunks)
```

## Fixed Size + Overlappinhg

```python
chunks =[]
fixed_size = 20
overlapping_size = 4
chunk_size = fixed_size - overlapping_size
for index in range(0, len(data), chunk_size): # 0 20
    chunk = data[index:index + fixed_size] #data[0:0+20] data[0:20]  #data[20:20+20]
    # print(chunk,len(chunk))
    chunks.append(chunk)
print(chunks)
```

## Sentence Based Chunking

```python
import re
sentences = [s.strip() for s in re.split(r'(?<=[.!?])\s+|\n+', data) if s.strip()]

sentence_chunks = []
current = ""

for s in sentences:
    candidate = (current + " " + s).strip() if current else s
    if len(candidate) <= chunk_size:
        current = candidate
    else:
        if current:
            sentence_chunks.append(current)
            current = ""
        if len(s) <= chunk_size:
            current = s
        else:
            # split long sentence into fixed-size pieces
            for i in range(0, len(s), chunk_size):
                sentence_chunks.append(s[i:i+chunk_size])
            current = ""

if current:
    sentence_chunks.append(current)

print(sentence_chunks)
```
