```python
def find_postag(tag):
    if tag.startswith('J'):
        return 'a'  # Adjective
    elif tag.startswith('N'):
        return 'n'  # Noun
    elif tag.startswith('V'):
        return 'v'  # Verb
    elif tag.startswith('R'):
        return 'r'  # Adverb
    else:
        return 'n'
```

```python
pos_tag_words = nltk.pos_tag(word_tokenize)
pos_tag_words
```

```python
for word,tag in pos_tag_words: #('running', 'VBG')
    tagged_word = lemmatizer.lemmatize(word, pos=find_postag(tag))
 
    print(tagged_word)
```
