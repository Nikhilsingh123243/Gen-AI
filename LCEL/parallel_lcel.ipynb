```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough,RunnableParallel
from langchain_community.document_loaders import PyPDFLoader,TextLoader
import os
```

```python
parser = StrOutputParser()
```

```python
import os
os.environ['Gemini_API2'] = os.environ["GEMINI_API_KEY"]
```

```python
import os
llm = ChatGoogleGenerativeAI(
    model = 'gemini-3.5-flash',
    api_key = os.environ["GEMINI_API_KEY"]
)
```

## LCEL 1 - Extracting the skills from the resume

```python
skill_prompt = ChatPromptTemplate.from_messages([
    ("system","Extracting the skills for the resume."),
    ("human","resume : {resume}")
    
]
)

skill_chain = skill_prompt | llm | parser
```

## LCEL 2 - Summarize the about project/exp from the resume

```python
summarize_prompt = ChatPromptTemplate.from_messages([
    ("system","Summarize the experience from the resume."),
    ("human","resume : {resume}")
]
)

summarize_chain = summarize_prompt | llm | parser
```

## LCEL 3 - Find the missing skills from the resume for Data science role

```python
missing_skills_prompt = ChatPromptTemplate.from_messages([
    ("system","For Data science role find the missing skills from the resume."),
    ("human","resume : {resume}")
])

missing_skills_chain = missing_skills_prompt | llm | parser
```

```python
final_chain = RunnableParallel(
    skills = skill_chain,
    summarize = summarize_chain,
    missing_skills = missing_skills_chain
)
```

```python
sample_resume = """Name: Jane Doe
Email: jane.doe@example.com
Phone: (555) 987-6543

Summary:
Data Scientist with 4 years of experience building predictive models, analyzing large datasets,
and translating business requirements into actionable analytics solutions.

Education:
- M.S. in Data Science, University of Somewhere, 2022
- B.S. in Statistics, State University, 2020

Experience:
Data Scientist | Acme Analytics | 2022 - Present
- Designed and deployed a customer churn prediction model that improved retention by 12%.
- Analyzed sales and marketing data to identify trends and recommend pricing strategies.
- Built automated dashboards in Tableau and Power BI for executive reporting.

Data Science Intern | NextGen Insights | 2021 - 2022
- Developed a machine learning pipeline for demand forecasting using Python and scikit-learn.
- Cleaned and processed 1M+ rows of transactional data from SQL databases.
- Created feature engineering workflows and evaluated model performance.

Skills:
- Programming: Python, R, SQL
- Machine Learning: regression, classification, clustering, NLP
- Tools: pandas, NumPy, scikit-learn, TensorFlow, PyTorch, Tableau, Power BI
- Data: data preprocessing, feature engineering, EDA, statistical analysis

Projects:
- Sales Forecasting Model: built a time series forecasting solution for monthly revenue predictions.
- Sentiment Analysis Pipeline: developed NLP model to analyze customer feedback and classify sentiment.

Certifications:
- IBM Data Science Professional Certificate
- Google Data Analytics Professional Certificate
"""

result = final_chain.invoke({"resume" : sample_resume})
print(result)
```
