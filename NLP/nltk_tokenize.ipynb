```python
res = ("Return a tokenized copy of text using NLTK's recommended word tokenizer currently an improved .Tree bank Word Tokenizer along with Punkt sentence Tokenizer for the specified language")
res.split()
```

```python
res.split(".")
```

```python
import re 


re.findall(r'\s+',res)
re.sub(r'\s+','',res)
```

```python
re.sub(r'\s{2,}','',res)
```

```python
re.sub(r'[^a-zA-Z0-9]','',res) #this pattern is used to remove special characters
```
