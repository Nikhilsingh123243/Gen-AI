```python
import nltk
from nltk.stem import WordNetLemmatizer
```

```python
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
tokens = ['running', 'cars', 'geese', 'better', 'happier', 'leaves', 'fishing', 'studies', 'children', 'mice']
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
for word in tokens:
    print(word,lemmatizer.lemmatize(word,pos='a'))
```

```python
nltk.pos_tag(["running"])
```

```python
nltk.pos_tag(tokens)

'''
N = 'n'  # Noun
V = 'v'  # Verb
JJ = 'a'  # Adjective
RBR = 'r'  # Adverb

'''
```

```python
def find_postag(tag):
    if tag.startswith('J'):
        return 'a'  # Adjective
    elif tag.startswith('N'):
        return 'n'  # Noun
    elif tag.startswith('V'):
        return 'v'  # Verb
    elif tag.startswith('R'):
        return 'r'  # Adverb
    else:
        return 'n'
```

```python
pos_tag_words = nltk.pos_tag(tokens)
pos_tag_words
```

```python
for word,tag in pos_tag_words: #('running', 'VBG')
    tagged_word = lemmatizer.lemmatize(word, pos=find_postag(tag))
 
    print(tagged_word)
```
