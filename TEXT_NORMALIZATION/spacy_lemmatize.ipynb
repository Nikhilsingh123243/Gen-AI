```python
import spacy
nlp = spacy.load("en_core_web_sm")

tokens = nlp('''Apply the Pipeline to some text.
             The text can span multiple sentences, 
             and can contain arbitrary whitespace.
             Alignment into the original string is Preserved.''')
print(type(tokens),tokens)
```

```python
lemmatize_tokens = [token.lemma_ for token in tokens  if not token.is_stop]
# for token in tokens:
    # print(type(token))
    # print(token.lemma_,token.is_stop)
print(lemmatize_tokens)
print(' '.join(lemmatize_tokens))
```

```python
text = ''' Bengaluru to Chennai, flight price is 5,000 Sql and this special offer is valid till 31st December 2025 on https://www.airline.com. '''
tokens = nlp(text)
```

```python
#print(type(tokens.ents))
for ent in tokens.ents:
    print(ent.text,ent.label_)
```

```python
for token in tokens:
    print(token.pos_,token.lemma_,token.ent_type_)
```
