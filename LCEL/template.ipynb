```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_google_genai import ChatGoogleGenerativeAI
```

## LCEL

## Step - 1 => prompt tempelate

```python
prompt_template = ChatPromptTemplate.from_messages([
    ('system',"you're a helpful assistant ,detect the native language from the given prompt"),
    ('ai','Expected Output : {language}')

])
```

## Step - 2 => LLM

```python
import os
llm = ChatGoogleGenerativeAI(
    model='gemini-3.5-flash',
    api_key= os.environ["GEMINI_API_KEY"],
    temperature=0.1
)
```

##

## step 3 = > Parser

```python
parser = StrOutputParser()
```

```python
language_chain = prompt_template | llm | parser

# response = language_chain.invoke(''' python ek easy subject hai , isme aapko programming ke basic cocepts sikhaye jaate hai, 
#  Python ka syntax simple aur readable hai, jo beginners ke liye perfect hai.''')

response = language_chain.invoke(''' என்று பெயர். ஒரே மாதிரியான கருétஐ அல்லது தகவலைச் சொல்லும் பல.''')
print(response)

```
