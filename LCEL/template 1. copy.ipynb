```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_google_genai import ChatGoogleGenerativeAI
import os
```

```python
import os
llm = ChatGoogleGenerativeAI(
    model = 'gemini-3.5-flash',
    api_key = os.environ["GEMINI_API_KEY"],
    temperature = 0.2
)
```

### LCEL1

```python
lang_detect_prompt_template = ChatPromptTemplate.from_messages(
    [
    ('system',
     "You 're a helpful assistant"
    "detect the native language from the given prompt"
    "Return only detected language"),
    ("human","{text}")
    ]
)
```

```python
parser =StrOutputParser()
```

```python
language_detect_chain = lang_detect_prompt_template | llm |parser
```

```python
translate_prompt_template = ChatPromptTemplate.from_messages(
    [
    ('system',
     "You 're a helpful assistant"
    "Detect the language from the given text and Translate  to English Language"
    "Return Detected Language and Translated text"),
    ("human","{text}")
    ]
)
```

```python
translated_chain = translate_prompt_template | llm | parser
```

```python
summarize_chain = summary_prompt_template|llm|parser
```
