```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_google_genai import ChatGoogleGenerativeAI
import os
```

## LCEL 1


## Step 1=> Prompt_Template

```python
lang_detect_prompt_template = ChatPromptTemplate.from_messages(
    [
        ("system",
         "you are a helpful assistant."
         "Detect the Native Language from the given prompt"
         "Return Only Detected Language and given text "),
        ("human","{text}")
        
    ]
)
```

## Step 2=> LLM

```python
import os
llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash", # gemini-3.6-flash,gemini-2.5-pro # os.enviorn
    api_key= os.environ["GEMINI_API_KEY"],
    temprature=0.0
)
```

## Step 3=> Parser

```python
parser = StrOutputParser()
```

```python
language =lang_detect_prompt_template | llm | parser
```
